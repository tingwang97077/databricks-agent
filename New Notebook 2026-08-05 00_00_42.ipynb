## Requirements

### 1. Create the Lakebase schema
**Create at least these two related tables:**

**tickets**

`ticket_id`
`title`
`status`
`created_by`
`created_at`

**ticket_messages**

`message_id`
`ticket_id`
`message_text`
`author`
`created_at`

The ticket_messages.ticket_id column must reference a ticket in the tickets table.
You may add additional columns or tables.

```python
%sql
create table tickets(
  ticket_id varchar(50) not null,
  title varchar(50) not null,
  status varchar(50) not null,
  created_by varchar(50) not null,
  created_at timestamp not null,
  primary key(ticket_id)
);

create table ticket_messages(
  message_id varchar(50) not null,
  ticket_id varchar(50) not null,
  message_text varchar(256),
  author varchar(50),
  created_at timestamp not null,
  primary key(message_id),
  foreign key(ticket_id) references tickets
);
```
